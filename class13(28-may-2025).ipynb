```python
#Task: Print the city having unique characters
```

```python
cities = ['Karachi', 'Lahore', 'Quetta','Dadu', 'Murre']
```

```python
for city in cities:
    lower_city = city.lower()
    for char in lower_city:
#         print(city)
        if lower_city.count(char) != 1:
            break
    else:
        print(city)
```

```python
for city in cities:
    for char in city.lower():
#         print(city)
        if city.lower().count(char) != 1:
            break
    else:
        print(city)
```

```python
for city in cities:
    empty_set = set()
    for char in city.lower():
        empty_set.add(char)
    if len(empty_set) == len(city):
        print(city)
```
