```python
class WeatherStation:
    def __init__(self, station):
        self.__station = station
        self.__observations = []
    
    def add_observation(self, observation):
        self.__observations.append(observation)

    def latest_observation(self):
        if self.__observations:
            print(f"The latest observation of {self.__station} is {self.__observations[-1]}")
        else:
            print("''")
    
    def number_of_observations(self):
        print(f"There are {len(self.__observations)} observation")
```

```python

station = WeatherStation("Houston")
```

```python
station.latest_observation()
```

```python
station.add_observation('Rain day')
```

```python
station.latest_observation()
```

```python
station.add_observation('Sunny')
```

```python
station.latest_observation()
```

```python
station.number_of_observations()
```

```python
class Series:
    def __init__(self, title, number, genres):
        self.title = title
        self.number = number
        self.genres = genres

    def dislay_info(self):
        print(f'''{self.title}: ({self.number} seasons)
genres: {', '.join(self.genres)}
no reviews  ''')
```

```python
dexter = Series("Dexter", 8, ["Crime", "Drama", "Mystery", "Thriller"])
```

```python
dexter.dislay_info()
```
