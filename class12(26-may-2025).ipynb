Iterate over a list (cities) and only print that city whose length is <= 4

```python
for city in cities:
    
    if len(city) > 4:
        print(city)
```

```python
even = []
odd = []

for num in range(1,21):
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
        
print(f'Even Numbers: {even}')
print(f'Odd Numbers: {odd}')
```

```python
for city in cities:
    for char in city:
        count = 0
        if city.count(char) == 1:
            count+=1
            if len(city) == count:
                print(city)
            continue
        else:
            break

```

```python
for city in cities:
    empty_set = set()    
    for char in city.lower():
        empty_set.add(char)
    if len(empty_set) == len(city):
        print(city)
```
